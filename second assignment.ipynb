```python
data = [12, 15, 12, 18, 20, 25, 12, 18, 30]
n = len(data)

# 1. Calculate Mean
mean = sum(data) / n

# 2. Calculate Median
sorted_data = sorted(data)
mid = n // 2

if n % 2 == 0:
    # If even, average the two middle numbers
    median = (sorted_data[mid - 1] + sorted_data[mid]) / 2
else:
    # If odd, take the middle number
    median = sorted_data[mid]

# 3. Calculate Mode
# Create a frequency dictionary
counts = {}
for x in data:
    counts[x] = counts.get(x, 0) + 1

# Find the highest frequency
max_freq = max(counts.values())

# Get all numbers that have that max frequency (handles ties)
mode = [key for key, val in counts.items() if val == max_freq]

# Output Results
print(f"Data: {data}")
print("-" * 20)
print(f"Mean:   {mean:.2f}")
print(f"Median: {median}")
print(f"Mode:   {mode}")
```

```python
# data = [[lower, upper, frequency]]
data = [[0, 10, 2], [10, 20, 7], [20, 30, 10]]

# 1. Mean
sum_fx = sum(f * (l + u) / 2 for l, u, f in data)
n = sum(f for l, u, f in data)
mean = sum_fx / n

# 2. Median
h = data[0][1] - data[0][0] # Class width
cf, cf_prev = 0, 0
for i, (l, u, f) in enumerate(data):
    cf += f
    if cf >= n / 2:
        median = l + ((n/2 - cf_prev) / f) * h
        break
    cf_prev = cf

# 3. Mode
max_f = max(f for l, u, f in data)
idx = [f for l, u, f in data].index(max_f)
l, u, f1 = data[idx]
f0 = data[idx-1][2] if idx > 0 else 0
f2 = data[idx+1][2] if idx < len(data)-1 else 0
mode = l + ((f1 - f0) / (2*f1 - f0 - f2)) * h

print(f"Mean: {mean}, Median: {median}, Mode: {mode}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample Data: Student Grades Categories
labels = ['A', 'B', 'C', 'D', 'F']
sizes = [15, 30, 35, 12, 8]  # Percentage or count

# Sample Data: Test Scores for Ogive
scores = [55, 62, 65, 71, 78, 82, 85, 88, 92, 95]
frequencies = [2, 5, 8, 12, 10, 7, 4, 3, 2, 1]
cumulative_freq = np.cumsum(frequencies)

# Create a figure with two subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# --- Plot 1: Pie Chart ---
ax1.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140, colors=plt.cm.Paired.colors)
ax1.set_title('Distribution of Grades (Pie Chart)')

# --- Plot 2: Ogive (Cumulative Frequency Graph) ---
ax2.plot(scores, cumulative_freq, marker='o', linestyle='-', color='teal')
ax2.set_title('Cumulative Frequency of Scores (Ogive)')
ax2.set_xlabel('Score Threshold')
ax2.set_ylabel('Cumulative Number of Students')
ax2.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import skew, skewnorm

# Set seed for reproducibility
np.random.seed(42)

# Generate synthetic data
n_samples = 1000
# 1. Symmetric (Normal)
symmetric_data = np.random.normal(loc=50, scale=10, size=n_samples)
# 2. Positively Skewed (Right-skewed)
positive_skew_data = skewnorm.rvs(a=10, loc=20, scale=10, size=n_samples)
# 3. Negatively Skewed (Left-skewed)
negative_skew_data = skewnorm.rvs(a=-10, loc=80, scale=10, size=n_samples)

datasets = [("Symmetric", symmetric_data), 
            ("Positive Skew", positive_skew_data), 
            ("Negative Skew", negative_skew_data)]

# Plotting
fig, axes = plt.subplots(2, 3, figsize=(18, 10))

for i, (label, data) in enumerate(datasets):
    # Calculate stats for annotation
    d_mean, d_median = np.mean(data), np.median(data)
    
    # Histogram + KDE (Top Row)
    sns.histplot(data, kde=True, ax=axes[0, i], color='skyblue', stat="density")
    axes[0, i].axvline(d_mean, color='red', linestyle='--', label=f'Mean: {d_mean:.1f}')
    axes[0, i].axvline(d_median, color='green', linestyle='-', label=f'Median: {d_median:.1f}')
    axes[0, i].set_title(f"{label} (Skew: {skew(data):.2f})")
    axes[0, i].legend()
    
    # Box plot (Bottom Row)
    sns.boxplot(x=data, ax=axes[1, i], color='lightgreen')
    axes[1, i].set_title(f"Box Plot: {label}")

plt.tight_layout()
plt.show()
```
